```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fig,ax = plt.subplots(figsize=(8, 5), dpi=100)
ax.plot(data["host_id"],data["price"])
ax.set_title("Title")
ax.set_xlabel("host_id")
ax.set_ylabel("price")
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8, 5), dpi=100)
ax.bar(data["availability_365"],data["price"])
ax.set_title("Title")
ax.set_xlabel("availability_365")
ax.set_ylabel("price")
plt.show()
```

```python

fig,ax = plt.subplots(figsize=(8, 5), dpi=100)
ax.barh(data["availability_365"],data["price"])
ax.set_title("Title")
ax.set_xlabel("availability_365")
ax.set_ylabel("price")
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8, 5), dpi=100)
ax.bar(data["availability_365"],data["price"],color=['#DE7272',"#31AAA9"])
ax.set_title("Title")
ax.set_xlabel("availability_365")
ax.set_ylabel("price")
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8, 5), dpi=100)
plot_data=data[["availability_365","price"]]
ax.errorbar(plot_data["availability_365"],plot_data["price"],fmt="o",capsize=4,alpha=0.5,color='#DE7272')
ax.set_title("Title")
ax.set_xlabel("availability_365")
ax.set_ylabel("price")
plt.show()
```
